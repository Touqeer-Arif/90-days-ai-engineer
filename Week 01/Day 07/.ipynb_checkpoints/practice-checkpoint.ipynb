```python
#Handle invalid number input.

try:
    number = int(input("Enter the number:"))
except ValueError:
    print("Enter an interger.")
else:
    print(number)
```

```python
#Prevent division by zero.

try:
    number1 = int(input("Enter the first number:"))
    number2 = int(input("Enter the second number:"))
    result = number1/number2
except ValueError:
    print("Enter a interger.")
except ZeroDivisionError:
    print("Cannot Divide by Zero.")
else:
    print(result)
```

```python
#Ask for age,Raise an exception if age < 0.

try:
    age = int(input("Enter the age: "))
    if age <= 0:
        raise ValueError("Age cannot be negative.")
except ValueError as e:
    if str(e) == "Age cannot be negative.":
        print(e)
    else:
        print("Please enter a valid integer.")
else:
    print(age)
```

```python
#Import math.Calculate:Square root,Power,Factorial

import math

print(f"Square root of 25 : {math.sqrt(25)}")
print(f"4 raised to the power 2 : {math.pow(4,2)}")
print(f"Factorial of 5 : {math.factorial(5)}")
```

```python
#Import random.Generate: Random number,Random choice from a list.

import random

number = random.randint(1,100)
print(f"Random Number : {number}")

fruits = ["apples","oranges",";bananas","grapes","strawberry"]
random = random.choice(fruits)
print(f"Randomm Fruit : {random}")
```

```python
#Import datetime.Display:Current date,Current time

import datetime

now = datetime.datetime.now()
print(f"Date : {now.date()}")
print(f"Time : {now.time()}")
```
