```python
#Print numbers from 1 to 100.
for i in range(1,101):
    print(i)
```

```python
#Print only even numbers from 1 to 100.
for i in range(1,101):
    if i % 2 != 0:
        continue
    print(i)
```

```python
#Print only odd numbers from 1 to 100.
for i in range(1,101):
    if i % 2 == 0:
        continue
    print(i)
```

```python
#Calculate the sum of numbers from 1 to 100.
count = 1
total = 0
while count <= 100:
    total += count
    count += 1
print(total)
```

```python
#Display the multiplication table of any number entered by the user.
number = int(input("Enter the number:"))
for i in range(1,11):
    print(number*i)
```

```python
#Reverse counting from 20 to 1.
for i in range(20,0,-1):
    print(i)
```

```python
#Count how many vowels are in a word entered by the user.
word = str(input("Enter the word:"))
vowels = "aeiouAEIOU"
count = 0
for char in word:
    if char in vowels:
        count += 1
print(count)
```

```python
# Find the factorial of a number
num = int(input("Enter a number: "))
factorial = 1
if num < 0:
    print("Factorial does not exist for negative numbers.")
else:
    for i in range(1, num + 1):
        factorial *= i
    print("Factorial of", num, "is", factorial)
```

```python
"""
Print a simple star pattern:

*
**
***
****
*****
"""

for i in range(1,6):
    print(i*"*")
```

```python
"""
Print the reverse pattern:

*****
****
***
**
*
"""
for i in range(5,0,-1):
    print(i*"*")
```
