```python
number = int(input("Enter the Number:"))
if number > 50:
    print("Number is greater than 50.")
elif number == 50:
    print("Number is equal to 50.")
else:
    print("Number is smaller than 50.")
```

```python
number = int(input("Enter the Obtained Number:"))
grade = str(input("Enter the Obtained Grade:"))

if (number >= 80 and number <= 100) and grade == "A":
    print("Excellent Performance")
elif (number >= 70 and number <= 79) and grade == "B": 
    print("Good Performance")
elif (number >= 60 and number <= 69) and grade == "C":
    print("Need Improvement")
elif (number >= 50 and number <= 59) and grade == "D":
    print("Poor Performance")
elif (number >= 0 and number <= 49) and grade == "NA":
    print("Fail")
else:
    print("Enter the right obtained numbers or obtained grade.")
```

```python
#Arthematic Operations

a = int(input("Enter the first number:"))
b = int(input("Enter the second number:"))
c = a+b
d = a-b
e = a//b #The output will be interger datatype
f = a/b #The output will be a float datatype
g = a*b 
print(f"The addition of the numbers give {c}.")
print(f"The Subtraction of the numbers give {d}.")
print(f"The Division of the numbers give {e}.")
print(f"The Division of the numbers give {f}.")
print(f"The Multiplication of the numbers give {g}.")
```

```python
#Temperature Conveter

temp = int(input("Enter the Celsius Temperature:"))
result = (temp * 9/5) + 32
print(f"The Temperature in Fahrenheit will be {result}.")
```

```python
#Rectangle Area

length = int(input("Enter the length of the Rectangle:"))
width = int(input("Enter the width of the Rectangle:"))
area = length * width 
print(f"The area of the Rectangle will be {area}.")
```

```python
#Rectangle Parameter

length = int(input("Enter the length of the Rectangle:"))
width = int(input("Enter the width of the Rectangle:"))
parameter = 2*(length+width) 
print(f"The parameter of the Rectangle will be {parameter}.")
```

```python
#Circle Area

import math
radius = int(input("Enter the radius of the Circle:"))
area = math.pi*radius**2
print(f"The area of the circle will be {area}.")
print(f"The area of the circle will be {area:.2f}.") #The ouput will include result upto 2 decimal place
```

```python
#Body Mass Index(BMI) Calculator

weight = int(input("Enter the weight in Kilograms:"))
height = float(input("Enter the Height in meters:"))
BMI =  weight/height**2
print(f"The BMI of the individual is {BMI}.")
```

```python
#Second Calculator

seconds = int(input("Enter the number of seconds:"))
hours = seconds // 3600
minutes = (seconds % 3600) // 60
remaining_seconds = seconds % 60
print(f"Hours: {hours}")
print(f"Minutes: {minutes}")
print(f"Seconds: {remaining_seconds}")
```

```python
#current date: 21 July,2026
#current market value:  1 PKR = 0.0036 USD

amount = float(input("Enter the amount in PKR:"))
exchange_rate = 0.0036 # 1 PKR = 0.0036 USD
USD = amount * exchange_rate
print(f"The amount in USD is: {USD}")

#Vice Versa
amount2 = float(input("Enter the amount in USD:"))
exchange_rate = 0.0036 # 1 PKR = 0.0036 USD
PKR = amount2 / exchange_rate
print(f"The amount in PKR is: {PKR}")
```

```python
#Student Profile Generator

full_name = str(input("Enter your Full Name:"))
father_name = str(input("Enter your Father's Name:"))
uni_name = str(input("Enter your University Name:"))
degree_name = str(input("Enter your Degree Name:"))
semester_no = int(input("Enter your Semester Number:"))
age = int(input("Enter your Age:"))
cgpa = float(input("Enter your current CGPA:"))
city = str(input("Enter your Current City of Residence:"))
country = str(input("Enter your Current Country of Residence:"))

if cgpa >= 3.5:
    performance = "Outstanding Performance"
elif cgpa >= 3.0:
    performance = "Good Performance"
else:
    performance = "Needs Improvement"

semester_left = 8 - semester_no
if semester_left == 0:
    year_left = "Your Degree is Completed"
elif semester_left == 1:
    year_left = "Half of an Year"
elif semester_left == 2:
    year_left = "An Year"
elif semester_left == 3:
    year_left = "One and a Half Year"
elif semester_left == 4:
    year_left = "Two Year"
elif semester_left == 5:
    year_left = "Two and a Half Year"
elif semester_left == 6:
    year_left = "Three Year"
elif semester_left == 7:
    year_left = "Three and a Half Year"
print()
print("         STUDENT PROFILE")
print()
print(f"Student Name : {full_name}")
print(f"Father Name  : {father_name}")
print(f"University   : {uni_name}")
print(f"Degree       : {degree_name}")
print(f"Semester     : {semester_no}")
print(f"Age          : {age}")
print(f"CGPA         : {cgpa}")
print(f"City         : {city}")
print(f"Country      : {country}")
print(f"Performance  : {performance}")
print(f"Years Left   : {year_left}")
```
