```python
#Arthematic Operations

a = int(input("Enter the first number:"))
b = int(input("Enter the second number:"))
c = a+b
d = a-b
e = a//b #The output will be interger datatype
f = a/b #The output will be a float datatype
g = a*b 
print(f"The addition of the numbers give {c}.")
print(f"The Subtraction of the numbers give {d}.")
print(f"The Division of the numbers give {e}.")
print(f"The Division of the numbers give {f}.")
print(f"The Multiplication of the numbers give {g}.")
```

```python
#Temperature Conveter

temp = int(input("Enter the Celsius Temperature:"))
result = (temp * 9/5) + 32
print(f"The Temperature in Fahrenheit will be {result}.")
```

```python
#Rectangle Area

length = int(input("Enter the length of the Rectangle:"))
width = int(input("Enter the width of the Rectangle:"))
area = length * width 
print(f"The area of the Rectangle will be {area}.")
```

```python
#Rectangle Parameter

length = int(input("Enter the length of the Rectangle:"))
width = int(input("Enter the width of the Rectangle:"))
parameter = 2*(length+width) 
print(f"The parameter of the Rectangle will be {parameter}.")
```

```python
#Circle Area

import math
radius = int(input("Enter the radius of the Circle:"))
area = math.pi*radius**2
print(f"The area of the circle will be {area}.")
print(f"The area of the circle will be {area:.2f}.") #The ouput will include result upto 2 decimal place
```

```python
#Body Mass Index(BMI) Calculator

weight = int(input("Enter the weight in Kilograms:"))
height = float(input("Enter the Height in meters:"))
BMI =  weight/height**2
print(f"The BMI of the individual is {BMI}.")
```

```python
#Second Calculator

seconds = int(input("Enter the number of seconds:"))
hours = seconds // 3600
minutes = (seconds % 3600) // 60
remaining_seconds = seconds % 60
print(f"Hours: {hours}")
print(f"Minutes: {minutes}")
print(f"Seconds: {remaining_seconds}")
```
