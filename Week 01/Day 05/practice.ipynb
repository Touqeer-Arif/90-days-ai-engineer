```python
#Ask the user for five numbers.

"""
Store them in a list.

Print:

Maximum
Minimum
Sum
Average
"""

list = []
for i in range(1,6):
    num = int(input("Enter the number:"))
    list.append(num)
print()
print(list)
print(f"Max Number : {max(list)}")
print(f"Min Number : {min(list)}")
print(f"Sum of Numbers : {sum(list)}")
print(f"The Average of Numbers : {sum(list)/len(list)}")
```

```python
#Sort a list in ascending and descending order.

list = [12,34,67,89,53,34,22,78]
list.sort()

print("Ascending Order:")
print(list)
print()
print("Descending Order:")
list.reverse()
print(list)
```

```python
#Create two sets.

"""
Display:

Union
Intersection
Difference
"""

set1 = {12,345,45,56,78,98}
set2 = {56,67,784,45,78,92}

print(f"Union : {set1.union(set2)}")
print(f"Intersection : {set1.intersection(set2)}")
print(f"Difference : {set1 - set2}")
```

```python
#Create a dictionary representing a student.Display each key and value.

dic = {
    "name" : "Harry",
    "age" : 19,
    "CGPA" : 3.45
}

for key,value in dic.items():
    print(f"{key} : {value}")
```

```python
#Allow the user to update the student's CGPA.

dic = {
    "name" : "Harry",
    "age" : 19,
    "CGPA" : 3.45
}


dic.update(CGPA = 3.70)

for key,value in dic.items():
    print(f"{key} : {value}")
```

```python
#Count how many times each word appears in a sentence using a dictionary.

sentence = input("Enter the sentence:")
word_count = {}

words = sentence.split()
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print(word_count)
```
