```python
#Convert a sentence to uppercase.

sentence = input("Enter the sentence:")

upper = sentence.upper()

print(upper)
```

```python
#Convert a sentence to lowercase.

sentence = input("Enter the sentence:")

lower = sentence.lower()
print(lower)
```

```python
#Count vowels in a sentence.

sentence = input("Enter the sentence:")
count_a = sentence.count("a")
count_e = sentence.count("e")
count_i = sentence.count("i")
count_o = sentence.count("o")
count_u = sentence.count("u")

total = count_a+count_e+count_i+count_o+count_u
print(f"Total Number of Vowels : {total}")
```

```python
#Reverse a string.

string = str(input("Enter the string:"))

print(f"Reverse : {string[::-1]}")
```

```python
#Check whether a string is a palindrome.

string = input("Enter the string:")

if string[::-1] == string:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
#Count words in a sentence.

sentence = input("Enter the sentence:")
words = sentence.split()
print(f"Total number of words in a sentence: {len(words)}")
```

```python
#Replace one word with another.

text = "I had always loved you Mikasa."
replace = text.replace("loved","Hated")
print(replace)
```

```python
#Create a file called:notes.txt , Write three lines into it.

with open("notes.txt","w") as file:
    file.write("Hello Mike.")
    file.write("\nHello John.")
    file.write("\nHello Rice.")
```

```python
#Read and display the contents of notes.txt.

with open("notes.txt","r") as file:
    data = file.read()

print(data)
```

```python
#Append another line to the same file.

with open("notes.txt","a") as file:
    file.write("\nHello Michael.")
```
