```python
#Check whether a number is:
#Positive
#Negative
#Zero

number = int(input("Enter the number:"))
if number > 0:
    print("The number is a Positive Number.")
elif number < 0:
    print("The number is a Negative Number.")
elif number == 0:
    print("The number is a Zero Number.")
else:
    print("Wrong input.")
```

```python
#Check if a number is:
#Even
#Odd

number = int(input("Enter the number:"))
if number % 2 == 0:
    print("The number is a Even Number.")
elif number % 2 != 0:
    print("The number is a Odd Number.")
else:
    print("Wrong Input")
```

```python
#Find the larger number between two numbers.

number1 = int(input("Enter the first number:"))
number2 = int(input("Enter the second number:"))
if number1 > number2:
    print("The First number is greater than Second number.")
elif number2 > number1:
    print("The Second number is greater than First number.")
elif number1 == number2:
    print("Both the number are equal.")
else:
    print("Wrong Input.")
```

```python
#Find the largest among three numbers.

number1 = int(input("Enter the first number:"))
number2 = int(input("Enter the second number:"))
number3 = int(input("Enter the third number:"))
if number1 > number2 and number1 > number3:
    print("The First number is the largest number.")
elif number2 > number1 and number2 > number3:
    print("The Second number is the largest number.")
elif number3 > number1 and number3 > number2:
    print("The Third number is the largest number.")
else:
    print("Wrong Input.")
```

```python
#Check whether a person can vote.
#Condition:
#Age ≥ 18

age = int(input("Enter your age:"))
if age >= 18:
    print("You can vote.")
else:
    print("You cannnot vote.")
```

```python
#Check if a character is:
#Vowel
#Consonant

vowels = ["a","e","i","o","u"]
character = str(input("Enter the character:"))
if character in vowels:
    print("The character is a vowel character.")
else:
    print("The character is a Consonant character.")
```

```python
#Password Checker
#Correct password:python123
#If correct:Access Granted
#Else:Access Denied

password = "python123"
password_e = str(input("Enter the password:"))
if password_e == password:
    print("Access Granted.")
else:
    print("Access Denied.")
```

```python
#Create a Simple Login

user = "harry"
password = "1234"
username = str(input("Enter the username:"))
password_e = str(input("Enter the password:"))
if username == user and password == password_e:
    print("Successfully Login.")
else:
    print("username or passord Incorrect.")
```

```python
#Simple Calculator

number1 = int(input("Enter the first number:"))
operation = str(input("Enter the operation(+,-,*,/):"))
number2 = int(input("Enter the second number:"))
if operation == "+":
    print(number1+number2)
elif operation == "-":
    print(number1-number2)
elif operation == "*":
    print(number1*number2)
elif operation == "/":
    print(number1/number2)
else:
    print("Wrong Input.")
```

```python
#Check if a year is a leap year.
year = int(input("Enter a year: "))
if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("Leap year")
else:
    print("Not a leap year")
```
