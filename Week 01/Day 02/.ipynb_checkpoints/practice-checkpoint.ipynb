```python
#Arithmetic Operators
number1 = 44
number2 = 32

print("1.Addition:",number1+number2) #Addition
print("2.Subtraction:",number1-number2) #Subtraction
print("3.Multiplication:",number1*number2) #Multiplication
print("4.Division:",number1/number2) #Division
print("5.Floor Division:",number1//number2) #Floor Division
print("6.Modulus:",number1%number2) #Modulus
print("7.Exponent(Power):",number1**number2) #Exponent(Power)
```

```python
#Comparison operators

x = 33
y = 78

print("1.",x == y) #Equal to
print("2.",x > y) #Greater than
print("3.",x < y) #Smaller than
print("4.",x >= y) #Greater than or equal
print("5.",x <= y) #Smaller than or equal
print("6.",x != y) #Not equal to
```

```python
#Logical Operators

a = 56
b = 96

print("and Operator:") #True if both conditions are True
print("1.",a >= 50 and b <= 100)
print("2.",a <= 50 and b >= 100)
print()
print("or Operator:") #True if at least one condition is True
print("1.",a >= 50 or b <= 100)
print("2.",a >= 50 or b >= 100)
print("3.",a <= 50 or b <= 100)
print("4.",a <= 50 or b >= 100)
print()
print("not Operator:") #Reverse Result
print(not a == 56)
print(not a == 96)
```

```python
#Assignment Operators

a = 5 #Assign 5 to variable "a"
a += 5 #Its same as a = a + 5
print(a)
print()
b = 4
b -= 4 #Its same as b = b - 4
print(b)
print()
c = 7
c *= 7 #Its same as c = c * 7
print(c)
print()
d = 8
d /= 8 #Its same as d = d / 8
print(d)
print()
e = 9
e //= 9 #Its same as e = e // 9
print(e)
print()
f = 10
f %= 10 #Its same as f = f % 10
print(f)
print()
g = 11
g **= 11 #Its same as g = g ** 11
print(g)
```

```python
#Membership Operators

list = ["yamal","rice","curullu","Pedri","saka"]
print("in Operator:") #value exists in the sequence
print("rice" in list)
print("neymar" in list)
print()
print("not in Operator:") #value does not exist in the sequence
print("saka" not in list)
print("mbappe" not in list)
```

```python
#Identity Operators

a = [56,778,89]
b = a
c = [56,778,89]
print("is Operator:") #Both variables refer to the same object in memory
print(a is b)
print(a is c)
print()
print("is not Operator:") #Variables refer to different objects
print(a is not b)
print(a is not c)
print()
print("Comparison:")
print(a == c)
print(a == b)
#Although as we can see that the value of a and c is equal.
#But the is operators works like are these exactly the same object
#and the equal operators checks if both have the same value
```

```python
#Conditional Statements

age = 20
if age >= 18:
    print("You can vote.")
```

```python
age = 15
if age >= 18:
    print("You can vote.")
else:
    print("You are a minor.")
```

```python
marks = 89

if marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("Fail")
```

```python
#Nested if statements

age = 23
license = False

if age >= 18:
    if license:
        print("You can Drive.")
    else:
        print("You need to get a license.")
else:
    print("You are a Minor.")
```

```python
#Multiple Coditions
age = 20
has_id = True

if age >= 18 and has_id: #More than one condition(both need to be statisfied)
    print("Entry Allowed.")
else:
    print("Entry Denied.")
    
```

```python
day = "Sunday"
if day == "Saturday" or day == "Sunday": #Atleast one condition need to be statisfied
    print("Weekend")
else:
    print("Working Days")
```

```python
logged_in = False
if not logged_in: #Reverse the boolean value
    print("Please Log in")
else:
    print("You are already Log in")
```

```python
name = ""
if name:
    print(name)
else:
    print("Name is Empty")
#Python takes empty and unfilled conditions as False
```
