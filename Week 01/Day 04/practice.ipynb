```python
#Create a function that finds the largest of three numbers.

def largest(a,b,c):
    if a > b and a > c:
        print(f"The largest number is : {a}")
    elif b > a and b > c:
        print(f"The largest number is : {b}")
    elif c > a and c > b:
        print(f"The largest number is : {c}")
    else:
        print("Wrong Input")

largest(104,78,456) 
```

```python
#Create a function that checks if a number is even or odd.

def check_number(a):
    if a % 2 == 0:
        print(f"The number {a} is a even Number.")
    elif a % 2 != 0:
        print(f"The number {a} is a odd Number.")
    else:
        print("Wrong Number.")

check_number(5)
```

```python
#Create a function that returns the factorial of a number.

def factorial(num):
    factorial = 1
    if num < 0:
        print("Factorial does not exist for a negative number.")
    else:
        for i in range(1,num + 1):
            factorial *= i
        print(f"Factorial of {num} is {factorial}.")

factorial(5)
```

```python
#Create a function that checks if a string is a palindrome.

def palindrome(word):
    string = word.lower().replace(" ", "")
    if string == string[::-1]:
        print("Palindrome")
    else:
        print("Not a Palindrome")
        
palindrome("Racecar")
```

```python
#Create a function that counts vowels in a string.

def count(word):
    total = 0
    vowels = "aeiouAEIOU"
    for char in word:
        if char in vowels:
            total += 1
    print(f"The total number of vowels in the word are : {total}")

count("pakistan")
```

```python
#Create a function that converts Celsius to Fahrenheit.

def temp(celcius):
    result = (celcius * 9/5) + 32
    print(f"The Temperature in Fahrenheit will be {result}.")

temp(32)
```
