```python
#Create a class: Student. Create two student objects.Display their names.

class Student:

    def __init__(self,name):
        self.name = name

    def display(self):
        print(f"My name is {self.name}.")
        
student1 = Student("Harry")
student2 = Student("Jack")

student1.display()
student2.display()
```

```python
#Create a class: Car.  Attributes: Brand, Model, Price. Print all information.

class Car:

    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price
        
    def display(self):
        print("Car's Brand :", self.brand)
        print("Car's Model :", self.model)
        print("Car's Price :", self.price)
        
car1=Car("Toyoto",2017,450000)
car2=Car("Honda",2022,670000)

car1.display()
print()
car2.display()
```

```python
#Create a class: Book. Method: display_info()

class Book:

    def __init__(self,name,author,published):
        self.name = name
        self.author = author
        self.published = published

    def display_info(self):
        print(f"The Book name is {self.name}.")
        print(f"It is written by {self.author}.")
        print(f"It was published in {self.published}.")

book1 = Book("The Philospher","Mike Michael",1999)
book2 = Book("The Grip","William Smith",2001)

book1.display_info()
print()
book2.display_info()
```

```python
#Create: Rectangle. Methods:Area, Perimeter

class Rectangle:

    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        area = self.length * self.width
        print(f"The Area of the Rectangle is {area}.")

    def perimeter(self):
        perimeter = 2 * (self.length + self.width)
        print(f"The Perimeter of the Rectangle is {perimeter}.")

rectangle1 = Rectangle(5,6)

rectangle1.area()
rectangle1.perimeter()
```

```python
#Create:Circle. Method: Calculate Area

import math

class Circle:

    def __init__(self,radius):
        self.radius = radius

    def area(self):
        area = math.pi*self.radius**2
        print(f"The area of the Circle is {area}.")

circle1 = Circle(2)

circle1.area()
```

```python
#Create: BankAccount. Methods:Deposit, Withdraw, Check Balance

class BankAccount:

    def __init__(self,balance = 0):
        self.balance = balance

    def deposit(self,amount):
        self.balance += amount
        print(f"Deposited Amount : ${amount}.")

    def withdrew(self,amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew Amount : ${amount}.")
        else:
            print("Insufficent Funds.")

    def check_balance(self):
        print(f"Current Balance : {self.balance}.")

account = BankAccount()

account.deposit(500)
account.withdrew(200)
account.check_balance()
        
```

```python
#Create: Employee. Display employee information.

class Employee:

    def __init__(self,name,id,role):
        self.name = name
        self.id = id
        self.role = role

    def display(self):
        print(f"Employee Name : {self.name}")
        print(f"Employee ID : {self.id}")
        print(f"Employee Role : {self.role}")

employee1 = Employee("June",456454,"HR")

employee1.display()
```

```python
#Create:Laptop. Method: Display Specifications

class Laptop:

    def __init__(self,brand,model,ram,storage):
        self.brand = brand
        self.model = model
        self.ram = ram
        self.storage = storage

    def display_specifications(self):
        print("Laptop Specifications:")
        print(f"Brand Name : {self.brand}")
        print(f"Model Name : {self.model}")
        print(f"RAM : {self.ram} GB")
        print(f"Storage : {self.storage} GB")

laptop1 = Laptop("Hp","core i5",8,12)

laptop1.display_specifications()
```

```python
#Create:Movie. Display movie information.

class Movie:

    def __init__(self,name,director,released):
        self.name = name
        self.director = director
        self.released = released

    def display(self):
        print(f"Movie Name : {self.name}")
        print(f"Director Name : {self.director}")
        print(f"Released Year : {self.released}")

movie1 = Movie("The Forbidden Land","Jack Williams",2017)

movie1.display()
```
